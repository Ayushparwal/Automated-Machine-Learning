```python
from pycaret.datasets import get_data
from pycaret.classification import *

# Load dataset
data = get_data('iris')

# Initialize setup
# This automatically does preprocessing and splits data internally
clf_setup = setup(data, target='species', session_id=123,  verbose=False)

# Compare multiple models and get the best one
best_model = compare_models()

# Create model (optional if you want a specific model)
model = create_model('rf')  # Random Forest example

# Tune the best model's hyperparameters
tuned_model = tune_model(best_model)

# Evaluate the tuned model (interactive plots)
evaluate_model(tuned_model)

# Finalize the model (train on full dataset)
final_model = finalize_model(tuned_model)

# Predict on new data (here using a sample of original data)
predictions = predict_model(final_model, data=data.sample(5))

print(predictions)
```
